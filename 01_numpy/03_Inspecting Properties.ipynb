```python
import numpy as np
```

```python
a = np.arange(1,10,2)
print(a)
print(type(a))
print(a.dtype)
# float
b = a.astype(float)
print(b)
print(type(b))
print(b.dtype)
# string
c = a.astype(str)
print(c)
print(c.dtype)
# boolean
d = a.astype(bool)
print(d)
print(d.dtype)
# others
for dtype in [np.int8, np.int16, np.float32, np.float16, np.float32, np.float64, np.bool_]:
    print(dtype, a.astype(dtype).dtype)
```

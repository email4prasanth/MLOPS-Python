```python
import numpy as np
```

```python
arr =  np.array([1, 2, 3, 4])
# copy
b = np.copy(arr)
print ("arr,b",arr,b)
b[1] = 100
print("No change in arr",arr)
# assignment
c = arr
c[1] = 100
print("arr changed",arr)
```

```python
arr = np.array([1, 2, 3, 4], dtype=np.int16)
b = arr.view(np.float16)
print(arr)
print(b)
```

```python
arr = np.random.randint(10, size = 10)
print(arr)
print(np.sort(arr))
```

```python
arr = np.random.randint(10, size = (2,10))
print(arr)
print(arr.sort(axis=1))
print(arr.flatten())
```

```python
a =  np.random.randint(9, size = (4,6))
print(a)
b = a.T
print(b)
C = a.reshape(2,12)
print(C)
del b
del C
d = a.resize((4,4))
print(d)
print("modified matrix \n",a)
e = np.resize(a,(10,7))
print(e)
```

```python
import numpy as np
```

```python
a =  np.random.randint(5, size = (3,4))
b =  np.random.randint(3, size = (5,6))
print(a)
print(b)
c = np.resize(a,(5,6))
print(c)
# add rows c at the end of b
con_row = np.concatenate((b, c), axis = 0)
print(con_row)
# add columns c at the end of b
con_col = np.concatenate((b, c), axis = 1)
print(con_col)
```

```python
# split
a =  np.random.randint(5, size = (3,4))
print(a)
# split into 3(rows) sub-arrays
split_arr = np.split(a,3) 
print(split_arr)
```

```python
b =  np.random.randint(3, size = (5,6))
print(b)
# split into 6(column) sub-arrays
hor_split = np.hsplit(b,6)
print(hor_split)
```

```python
c = np.array([[1,2,3,10],[4,5,6,12],[8,9,10,11],[14,15,16,2],])
print(c)
print(c[:,1:3]) #columns
print(c[0:2,:]) # rows
print(c[:,-1]) # last column
print(c[-1,:]) # last row
print(c[:,-3:-1]) # -3, -2 column
```

```python
c = np.random.randint(5, size=(3,15))
print("c\n",c)
# first three columns
print(f"first three columns \n {c[:,0:3]}")
# last three columns -1,-2,-3
print(c[:,-3:])
# only last column
print(c[:,-1])
```

```python
d = np.random.randint(5, size=(10,5))
print("d\n",d)
# firt 4 rows
print(f"firt 4 rows \n{d[0:4,:]}")
# last 4 columns
print(f"last 4 columns \n {d[-4:,:]}")
# last coulmn
print(f"last column \n{d[-1,:]}")
```

```python
# specific element
e = np.random.randint(5, size=(8,5))
print("e\n",e)

# last but three rows and last but 2 columns
print(f"last but three rows \n {e[-4:-1, :]}")
print(f"last but 2 columns \n {e[:, -3:-1]}")
print(f"last but three rows and last but 2 columns \n {e[-4:-1, -3:-1]}")
```

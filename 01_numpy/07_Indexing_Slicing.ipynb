```python
import numpy as np
```

```python
# Indexing
arr = np.random.randint(6, size = (4,5))
print(arr)
print(arr[2]) # 3rd row (elements of 3rd index)
print(arr[2][3]) # 3rd row, 4th column
print(arr[2,3])
```

```python
# Slicing
arr = np.random.randint(6, size = (5,5))
print(arr)
print(arr[1:4]) # Index of 1,2,3
print(arr[1:4,2]) # rows of 1,2,3 and column of 2
print(arr[1:4,2:3])
print(arr[:2]) # 0,1 rows
print(arr[3:]) # 3,4 rows
print(arr[:,3:]) # 3,4 columns
print(arr[:,1:3]) # 1,2 colums

```

```python
# Conditional stmts
arr = np.random.randint(6, size = (5,5))
print(arr)
print(arr<4)
```

```python
a = np.random.randint(4, size = (5,5))
b = np.random.randint(6, size = (5,5))
print(a)
print(~a)
print(b)
# and condition
print((a<2) & (b>3))
# or condition
print((a < 2) | (b > 3))
```

```python
arr = np.random.randint(5, size = (3,4))
print(arr)
print(arr[arr<3])
```

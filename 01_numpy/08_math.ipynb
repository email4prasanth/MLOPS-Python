```python
import numpy as np
```

```python
# scalar math
a = np.array([1,2,4])
print(a)
print(np.add(a,1))
print(np.subtract(a,2))
print(np.multiply(a,2))
print(np.divide(a,2))
print(np.divmod(a,2))
print(np.power(a,3))
print(np.exp(a))
print(np.log10(a))
print(np.log2(a))
```

```python
# vector math
a = np.random.rand(5,6)
print(a)
b =  np.random.randint(5, size = (5,6))
print(b)
print(np.add(a,b))
print(np.subtract(a,b))
print(np.multiply(a,b))
```

```python
a = np.random.rand(5,6)
print(a)
b =  np.random.randint(5, size = (5,6))
print(b)
print(np.divide(a,b))
print(np.power(a,b))
print(np.array_equal(a,b))
```

```python
a =  np.array([2,3,4])
print(np.sqrt(a))
print(np.square(a))
print(np.sin(a))
print(np.log(a))
b = np.array([2+3j])
print(b)
print(np.abs(b))
print(abs(b))
```

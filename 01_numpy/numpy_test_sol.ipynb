```python
# Q5: Given arr=[10,20,30,40,50]: slice index 1 to 3, replace values >30 with 0
arr=[10,20,30,40,50]
print(arr[1:4])
for i in range (len(arr)):
    if arr[i] >30:
        arr[i] = 0
print(arr)
# Corrected
arr = np.array([10,20,30,40,50])
arr[arr > 30] = 0
```

```python
# Q6: Given a=[1,2,3], b=[4,5,6]: multiply element-wise, then take sqrt
a=[1,2,3]
b=[4,5,6]
c = np.multiply(a,b)
print(f"element multiple:{c}")
d = np.sqrt(c)
print(f"Square root: {d}:")
```

```python
# Q7: Create 4x4 random array. Find mean and column-wise max.
random_array = np.random.randint(4,size = (4,4))
print(f"random_array: \n{random_array}")
# correct
print(random_array.mean())
print(random_array.max())

mean_array = np.zeros(4)
column_max = np.zeros(4)
for i in range(len(random_array)):
    mean_array[i] = np.mean(random_array[:,i])
    column_max[i] = np.max(random_array[:,i])
print(f"mean value:{mean_array}")
print(f"max value:{column_max}")

```

```python
# Q9: Given [2,4,6,8,10]: boolean mask >5, extract and square them.
Arr = np.arange(2,12,2)
print(Arr)
filter = Arr > 5
filter_array = Arr[filter]
sq = np.square(filter_array)
print(sq)
Arr[filter] = Arr[filter]**2
print(Arr)
```

```python
# Q10: arange(1,13): reshape to 3x4, get 2nd row, last column, subarray (2x2).
ar = np.arange(1,13)
print(ar,len(ar))
ar_reshape = ar.reshape(3,4)
print(ar_reshape)
# second row
print(f"second row: {ar_reshape[1,:]}")
print(f"last column: {ar_reshape[:,-1]}")
print(f"sub array 2x2:\n {ar_reshape[:2,:2]}")
```

```python
# Q11: Concatenate [[1,2],[3,4]] and [[5,6]] row-wise and column-wise.
a = [[1,2],[3,4]]
b = [[5,6]]
print(a,np.shape(a))
print(b,np.shape(b))
row_con = np.concatenate((a,b), axis = 0)
print(row_con)
```

```python
# Q12: Given [10,25,30,45,50,5]: filter 20–50, replace <20 with -1, find mean & std.
a = np.array([10,25,30,45,50,5])
for i in range(len(a)):
    if a[i]<20:
        a[i] = -1
print(a)
print(f"mean and standard deviation are: {np.mean(a), np.std(a)}")
filtered = a[(a >= 20) & (a <= 50)]
print(filtered)
```

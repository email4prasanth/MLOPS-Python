```python
 # Checks for null Values, Returns  Boolean Array
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, np.nan, 35],
    'Country': ['USA', 'Canada', 'UK']
}
df = pd.DataFrame(data)
df.isnull()
# Show only rows with missing values
df[df.isnull().any(axis=1)]
```

```python
# Opposite of  pd.isnull()
df['Age'].notnull()
```

```python
# Drops all rows that contain null values
df.dropna()
```

```python
#  Drops all columns that contain null values
df.dropna(axis=1)
```

```python
# Drops all rows have have less than  n  non-null values
df.dropna(axis = 0, thresh=30)
```

```python
# Replace nan with mean
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, np.nan, 35],
    'Country': ['USA', 'Canada', 'UK']
}
df = pd.DataFrame(data)
df['Age']=df['Age'].fillna(df['Age'].mean())
df
```

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, np.nan, 35],
    'Country': ['USA', 'Canada', 'UK']
}
df = pd.DataFrame(data)
# Mass renaming of columns
df.rename(columns=lambda x: x+"_1", inplace=True)
print(df)

df.rename(columns=lambda x: x.upper(), inplace=True)
print(df)

df.columns = [f"col_{i}" for i in range(len(df.columns))]
print(df)
```

```python
# selective renaming
df = df.rename(columns={'col_0' : 'Name'})
df
```

```python
#  Mass renaming of index
df.rename(index=lambda x : x+10, inplace=True)
df
```

```python
import pandas as pd

data = {
    'Company': ['A', 'B', 'C', 'D', 'E'],
    'Sector': ['Tech', 'Finance', 'Tech', 'Finance', 'Tech'],
    'Country': ['USA', 'USA', 'UK', 'UK', 'USA'],
    'Revenue': [100, 200, 150, 120, 180]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Group by Sector and calculate average revenue:
df.groupby('Sector')['Revenue'].mean()
```

```python
# Group by Sector + Country
df.groupby(['Sector', 'Country'])['Revenue'].mean()
```

```python
# Pivot Table
pd.pivot_table(df, 
               values='Revenue', 
               index='Sector', 
               columns='Country', 
               aggfunc='sum')
```

```python
import numpy as np
df.groupby('Sector').agg({'Revenue': 'mean'})
```

```python
df.select_dtypes(include='number').apply(np.mean)
```

```python
import pandas as pd

data = {
    'Age': [25, 30, 35, 40, 45],
    'Salary': [30000, 40000, 50000, 60000, 70000],
    'Experience': [1, 3, 5, 7, 9]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.describe())
print(f"mean:\n{df.mean()}")
print(f"correlation:\n{df.corr()}")
print(f"count:\n{df.count()}")
print(f"Max:\n{df.max()}")
```

```python
print(f"Min:\n{df.min()}")
print(f"Media:\n{df.median()}")
print(f"Standard Deviation:\n{df.std()}")
```

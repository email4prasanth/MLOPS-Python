```python
from sklearn import preprocessing, datasets
iris = datasets.load_iris()
```

```python
from sklearn.model_selection import train_test_split
X = iris.data[:,:3] # selecting sepal length, sepal width
y = iris.target
# split train and test data with 80:20
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=32)
print(X_train, X_test, y_train, y_test )
```

```python
# Dictionary to store models and their accuracies
models = {}
accuracies = []
```

```python
scalar = preprocessing.StandardScaler().fit(X_train)
X_train_scaled = scalar.transform(X_train)
X_test_scaled = scalar.transform(X_test)

from sklearn.metrics import accuracy_score
from sklearn import neighbors
knn_model = neighbors.KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train_scaled, y_train)
y_pred =  knn_model.predict(X_test_scaled)
acc_knn = accuracy_score(y_test, y_pred)
print(acc_knn)
models['K-Nearest Neighbors'] = knn_model
accuracies.append(('K-Nearest Neighbors', acc_knn, X_train_scaled, X_test_scaled, y_train, y_test))
```

```python
# linear Model
from sklearn.linear_model import LogisticRegression
lr_model = LogisticRegression()
lr_model.fit(X_train_scaled, y_train)
y_pred = lr_model.predict(X_test_scaled)
acc_lr = accuracy_score(y_test,y_pred)
print(acc_lr)
models['Logistic Regression'] = lr_model
accuracies.append(('Logistic Regression', acc_lr, X_train_scaled, X_test_scaled, y_train, y_test))
```

```python
# SVM
from sklearn.svm import SVC
svc_model = SVC()
svc_model.fit(X_train_scaled, y_train)
y_pred = svc_model.predict(X_test_scaled)
acc_svm = accuracy_score(y_test,y_pred)
print(acc_svm)
models['SVM'] = svc_model
accuracies.append(('SVM', acc_svm, X_train_scaled, X_test_scaled, y_train, y_test))
```

```python
# Tree
from sklearn.tree import DecisionTreeClassifier

tree_model = DecisionTreeClassifier(random_state=32)
tree_model.fit(X_train_scaled, y_train)
y_pred = tree_model.predict(X_test_scaled)
acc_tree = accuracy_score(y_test,y_pred)
print(acc_tree)
models['Decision Tree'] = tree_model
accuracies.append(('Decision Tree', acc_tree, X_train_scaled, X_test_scaled, y_train, y_test))
```

```python
import pandas as pd
# Create comparison table
results_df = pd.DataFrame([
    {'Model': name, 'Accuracy': acc}
    for name, acc, _, _, _, _ in accuracies
])

# Sort by accuracy
results_df = results_df.sort_values('Accuracy', ascending=False)
results_df['Rank'] = range(1, len(results_df) + 1)

# Display table
print("\n" + "="*60)
print("MODEL COMPARISON RESULTS")
print("="*60)
print(results_df.to_string(index=False))
print("="*60)

# Find best model
best_model_name = results_df.iloc[0]['Model']
best_accuracy = results_df.iloc[0]['Accuracy']
print(f"\n🏆 BEST MODEL: {best_model_name} with Accuracy = {best_accuracy:.4f} ({best_accuracy*100:.2f}%)")
```
